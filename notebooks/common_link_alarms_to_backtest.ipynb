# Link of Stream A and Stream B: Decisions on Detector Alarms

**Master's Thesis Project:** Seasonality-Aware Drift Detection System  
**Week 4:** the backtest of stream B takes its decisions at the alarms of a drift detector instead of on a weekly calendar. The detector is a black box that writes one file of alarms; the strategies that review the model on alarm are compared with the weekly ones on the same 263 series, and the cost of false alarms shows up in retrainings and checks on series without a change.

Two detectors run on the same series: the detector of stream A, `residual_conformal` (`ResidualConformalDriftDetector` of `src/detector` with the settings of its multi-series mode for a daily period; its residual of an hour comes from an STL fit on a block that ends up to 23 hours later, so an alarm is dated at the end of its block), and `adwin` (ADWIN on the residual of each hour from the median of the same hour of the last 4 weeks, which uses only the past). Both are calibrated per source on a control period without changes to one false alarm per 14 days. A conformal stand-in (`profile_conformal`) served until the stream A detector was connected; it has been removed.

Computed by `python -m src.recommendations.alarms` (alarms) and `python -m src.recommendations.run_backtest --alerts data/alerts/alerts.csv --advisors rules` (strategies on alarm); the results are read from `data/alerts/` and `data/processed/stream_b/`. The rows of the weekly strategies are the same as in `03_stream_b_acceptance_rule.ipynb`.

## 4. What the Alarms Change

Each weekly strategy against its version on alarm, per detector: retrainings saved, MASE after the change, the share of series where the version on alarm is not worse, and the reaction time.

```python
pairs = [("scheduled", "alarm_retrain"), ("scheduled_gate", "alarm_gate"), ("catalog_gate", "alarm_catalog_gate"),
         ("rules_gate", "alarm_rules_gate"), ("oracle_weekly", "alarm_oracle")]
weekly = runs[runs.detector.isna()].set_index(KEY + ["strategy"])
rows = []
for det in DETECTORS:
    on = runs[runs.detector == det].set_index(KEY + ["strategy"])
    for a, b in pairs:
        for (src, sc), g in on.xs(b, level="strategy").groupby(level=["source", "scenario"]):
            w = weekly.xs(a, level="strategy").loc[g.index]
            rows.append(dict(detector=det, source=src, scenario=sc, weekly=a, on_alarm=b, series=len(g),
                             retrains_saved=round(1 - g.retrains.mean() / w.retrains.mean(), 2) if w.retrains.mean() > 0 else np.nan,
                             mase_weekly=round(w.mase_post.mean(), 3), mase_on_alarm=round(g.mase_post.mean(), 3),
                             share_not_worse=round((g.mase_post <= w.mase_post + 1e-9).mean(), 2),
                             reaction_h_on_alarm=round(g.reaction_h.mean(), 0)))
pd.DataFrame(rows)
```

## 6. The Chain on One Series

The Milan seismic station (IV.MILN) around the measures of March 2020: the alarms of both detectors, the review at the first origin after an alarm, the diagnostics the advisor sees, its recommendation, the acceptance rule on the two weeks before the review, and the forecasts of the model with and without the chain.

```python
from src.data.frames import series_id, series_table
from src.recommendations.data_sources import seismic_series
from src.recommendations.llm_client import RuleAdvisor, diagnostics

(s,) = seismic_series("../data/raw/seismic_noise", stations=("IV.MILN",))
sid = series_id(s)
mine = read_alerts("../data/alerts/alerts.csv", series_table([s]))[sid]
stamps = pd.date_range(s["start"], periods=len(s["y"]), freq="1h")
print("change (first measures):", stamps[s["t_change"]])
for det, idx in mine.items():
    near = [i for i in idx if s["t_change"] - 2 * WEEK <= i < s["t_change"] + 2 * WEEK]
    print(f"{det}: {len(idx)} alarms after deployment; around the change:", ", ".join(str(stamps[i]) for i in near))
```

```python
det = "residual_conformal"
first = int(next(i for i in mine[det] if i >= s["t_change"]))
T = s["t_deploy"] + ((first - s["t_deploy"]) // DAY + 1) * DAY  # the first origin after the alarm
scale = mase_scale(s["y"][: s["t_deploy"]])
nothing = backtest(s["y"], s["F"], s["t_deploy"], STRATEGIES[0], s["base_features"])
ctx = diagnostics(s["y"], nothing["pred"], T, s["t_deploy"], scale, s["t_deploy"], alarm=dict(method=det, hours_ago=int(T - first)))
rec = RuleAdvisor().recommend(ctx)
print("review at", stamps[T], "| diagnostics:", {k: v for k, v in ctx.items() if k != "alarm"})
print("rules advisor:", rec.action, "-", rec.reason)
act = make_actor(s["F"], s["y"], s["base_features"])
model, _ = act(ACTIONS["retrain_full"], Model(s["base_features"]), s["t_deploy"])
if rec.action != "keep":
    print("acceptance rule:", check_action(ACTIONS[rec.action], model, s["y"], s["F"], nothing["pred"], T, scale, act))
```

```python
BY = {st.name: st for st in ALARM_STRATEGIES}
chain = {name: backtest(s["y"], s["F"], s["t_deploy"], BY[name], s["base_features"], alarms=mine[det], detector=det,
                        advisors={"rules": RuleAdvisor()})
         for name in ("alarm_retrain", "alarm_rules_gate")}
for name, r in chain.items():
    print(f"{name}: model changed at", [str(stamps[t]) for t in r["applied"]], "| picks:", r["picks"], "| retrains:", r["retrains"])
lo, hi = s["t_change"] - 2 * WEEK, s["t_change"] + 4 * WEEK
rows = np.arange(lo, hi)
days = (rows - s["t_change"]) / DAY
with plt.rc_context(STYLE):
    fig, ax = plt.subplots(figsize=(10, 4))
    ax.plot(days, s["y"][rows], color="#b8b7b1", lw=1.4, label="actual")
    ax.plot(days, nothing["pred"][rows], color=LINE["nothing"], lw=1.2, label=label_of("nothing"))
    for (name, r), color in zip(chain.items(), (LINE["scheduled"], LINE["oracle"])):
        ax.plot(days, r["pred"][rows], color=color, lw=1.2, label=label_of(name))
    for i in mine[det]:
        if lo <= i < hi:
            ax.axvline((i - s["t_change"]) / DAY, color="#52514e", lw=0.8, ls=":")
    ax.axvline(0, color="#0b0b0b", lw=1)
    ax.set_xlabel("days from the first measures")
    ax.set_ylabel("log10 noise level")
    ax.set_title(f"IV.MILN: alarms of {det} (dotted), decisions on alarm", loc="left", fontsize=10)
    ax.legend(loc="upper right", fontsize=8)
    plt.show()
```

## 8. Findings

All numbers are means over the series of a source and scenario (tables above), MASE after the change with the retrainings per series in brackets.

- **A detector's false alarm rate has to be calibrated per source.** Settings calibrated on the synthetic control series do not carry over: the multiplier of the stream A detector chosen there (4) gives 5.5 false alarms per 1000 hours on the national load and 7.4 on the seismic noise, and the ADWIN delta (0.003) gives 14.6 on the national load and 0.7 on the seismic noise, against a target of 2.98. Calibrated on a control period of each source, the stream A detector meets the target on the national load and the synthetic series (2.0 and 2.9 per 1000 hours) and is slightly above it on the seismic noise (3.0 with the largest multiplier of its grid, 8); ADWIN does not reach it on the national load even at its strictest setting (8.3).
- **Retraining on alarm saves retrainings at the price of a later reaction.** National load: 0.708 (8.6) with the stream A detector and 0.709 (18.4) with ADWIN against 0.696 (24) for weekly retraining; seismic noise: 1.369 (3.5) and 1.084 (4.4) against 0.998 (11); synthetic level shift: 1.178 (4.6) and 1.134 (6.1) against 1.091 (13). The difference between the detectors is the reaction time: on ADWIN alarms the model is changed 72 hours after the change on the national load, 128 hours on the seismic noise and 62 hours after the synthetic shift, on the alarms of the stream A detector 381, 267 and 543 hours.
- **On the national load the stream A detector matches ADWIN with half the retrainings; on the seismic noise it helps little.** It raises 11.5 alarms per series on the national load against 34.8 for ADWIN and reaches the same error. On the seismic noise retraining on its alarms (1.369) is closer to doing nothing (1.449) than to retraining on ADWIN alarms (1.084).
- **Late alarms cap what any advisor can do.** `alarm_oracle`, the best action in hindsight at the alarm times, is close to the weekly oracle with ADWIN (national load 0.636 against 0.607, synthetic shift 1.077 against 1.021, seismic noise 0.982 against 0.885) and far from it with the stream A detector (0.726, 1.159 and 1.386). The choice of the action still matters, but alarms that come weeks after the change leave little to choose.
- **The cost of false alarms is visible and small.** On the control series retraining on alarm changes the model 4.6 to 4.7 times per series in 14 weeks (one alarm every two to three weeks, as calibrated) against 13 weekly retrainings, with almost the same error (0.880 and 0.878 against 0.870); with the acceptance rule 0.85 to 0.88 times (scheduled_gate: 2.0). With doubled noise, retraining on alarm does no harm (1.669 and 1.670, the level of doing nothing) while weekly retraining does a little (1.673).
- **The acceptance rule on alarm reacts slowly.** `alarm_gate` and `alarm_catalog_gate` change the model 230 to 1200 hours after the change (one to seven weeks): the review at the first alarm after the change replays the action on two weeks that are mostly before the change and rejects it, and the next review waits for the next alarm. The check needs two weeks of evidence after the change; on a calendar it gets them at the next weekly review, on alarms only at the next alarm. The fix to try: keep a rejected proposal pending and re-check it daily (two fits a day for one action) until it passes or two weeks have gone by.
- **Advice without the check is as bad on alarm as on the calendar.** The rules advisor applied on alarm is worse than doing nothing on the national load (0.963 with the stream A detector and 1.180 with ADWIN against 0.787) and with doubled noise (1.80 to 1.82 against 1.669); with the check it returns to 0.82 to 0.84 and 1.67. This is the same protection the acceptance rule gave the weekly strategies.
- **ADWIN is the stronger trigger for this loop.** It reacts sooner after every change, which is what retraining on alarm needs; the stream A detector raises fewer alarms on the national load, where ADWIN cannot be brought down to the target rate, and saves half the retrainings there.
